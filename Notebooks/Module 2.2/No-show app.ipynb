**4. Initial data cleaning and transformation**

(Each person from the group)Choose a relevant column with one or multiple data issues and try to address them. For example: missing values, incorrect data type, mixed types of values, etc.

```python
# I need to remove negative age entries, >110 age entries. 
# I need to convert the "ScheduledDay" and "AppointmentDay" columns to datetime format
# Then remove any appointments that were scheduled after the appointment day.
# Remove negative ages
ma = ma[ma['Age'] >= 0]

# Remove ages above 110
ma = ma[ma['Age'] <= 110]

# Convert columns to datetime format
ma['ScheduledDate'] = pd.to_datetime(ma['ScheduledDate'])
ma['AppointmentDate'] = pd.to_datetime(ma['AppointmentDate'])

# Remove appointments scheduled after the appointment day
ma = ma[ma['ScheduledDate'] <= ma['AppointmentDate']]

print(f"Dataset shape after cleaning: {ma.shape}")
print(f"Remaining records: {len(ma)}")
```
