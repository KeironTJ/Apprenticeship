# Business Case

As a data engineering and AI team, we have been tasked by our client, "GlobaMart," a large-scale international retailer, to improve their supply chain and inventory management. GlobaMart currently uses outdated, subjective forecasting methods that often lead to two major problems:

1. **Overstocking**: We order too much of a product, leading to wasted inventory, storage costs, and potential product spoilage (especially for perishable goods).
2. **Understocking**: We don't order enough of a popular item, resulting in lost sales and frustrated customers.

Our goal is to develop a robust and scalable demand forecasting system that provides more accurate predictions of future sales. This will allow GlobaMart to optimise their inventory, reduce waste, and improve customer satisfaction. The project will serve as a proof of concept for the new AI-driven approach, which will be deployed across all stores in the future.

# Data

For this project, we will be using a dataset from the ongoing Kaggle competition, Store Sales - Time Series Forecasting. The data is provided by a large Ecuadorian grocery retailer, Corporación Favorita.

The dataset consists of several related files:

- train.csv: Contains the historical daily sales data, including the target variable sales.

- stores.csv: Provides metadata for each store, such as city, state, and store type.

- holidays_events.csv: Lists holidays and special events, a critical external factor for time series analysis.

- oil.csv: Contains daily oil prices, which can significantly impact the economy and sales in Ecuador.

- test.csv: The test data you will use to generate your final predictions.
