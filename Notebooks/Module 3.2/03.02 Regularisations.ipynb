```python
# import packages
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
pd.options.display.max_rows

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import root_mean_squared_error
from sklearn.model_selection import KFold, cross_val_score
```

```python
diamonds = sns.load_dataset('diamonds')
diamonds.head()
```

```python
X = diamonds[['carat', 'depth', 'table', 'x', 'y', 'z']]
y = diamonds['price']

X_train, X_test, y_train, y_test = train_test_split(X, 
                                                    y, 
                                                    test_size=0.3, 
                                                    random_state = 2021
                                                   )

model = LinearRegression()
model = model.fit(X_train, y_train)

y_predicted = model.predict(X_test)

print('Train r2 =', model.score(X_train, y_train))
print('Test r2 =', model.score(X_test, y_test))

print('MAE =', mean_absolute_error(y_test, y_predicted))
print('RMSE =', mean_squared_error(y_test, y_predicted))
```

```python
def cross_val_results(X, y, model = LinearRegression(), cv=5, scoring = 'r2'):
    kf = KFold(n_splits=10, shuffle=True, random_state=42)
    scores = cross_val_score(model, X, y, cv=kf, scoring='r2')
    print(f"Scores: {scores}")
    print(f"Mean: {scores.mean()}")
    print(f"Std: {scores.std()}")
    return scores
```

```python
cross_val_results(X, y, model)
```

### Select the features with Ridge, Lasso and ElasticNet models

#### Regularisators

Linear Regression is the most basic method from the regression family, where the model is not penalised for its choice of weights at all. However, this means that during the training process, if the model considers a particular feature important, it will assign a large weight to that feature. This can sometimes cause overfitting in smaller datasets.

Lasso, Ridge, and ElasticNet are methods from the linear regression family that penalize the model for its weights. They reduce model complexity and prevent over-fitting.

#### Ridge

In Ridge regression, the cost function is altered by adding a penalty equivalent to the square of the magnitude of the coefficients.

The penalty term lambda('alpha’ = 1.0(default) in Scikit Learn) regularises the coefficients such that if the coefficients take large values, the optimisation function is penalised.

The higher the alpha is, the bigger the penalty term, and therefore, the magnitude of the coefficients is reduced.

Therefore, Ridge regression shrinks the coefficients, which helps reduce model complexity and mitigate multicollinearity.

https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.Ridge.html

```python
from sklearn.linear_model import Ridge
```

```python
X = diamonds[['carat', 'depth', 'table', 'x', 'y', 'z']]
y = diamonds['price']

X_train, X_test, y_train, y_test = train_test_split(X, 
                                                    y, 
                                                    test_size=0.3, 
                                                    random_state = 2021
                                                   )
```

```python
# in scikit-learn lambda is specified as alpha, and by default it is eqaul to 1. 
# Larger values specify stronger regularization.

ridge_regr_model = Ridge(alpha=1)

ridge_regr_model.fit(X_train, y_train)
```

```python
ridge_regr_model.score(X_train, y_train)
```

```python
ridge_regr_model.score(X_test, y_test)
```

```python
cross_val_results(X, y, model=Ridge(alpha=1))
```

#### Lasso

Lasso is a modified linear regression method where the model is penalised for the sum of the absolute values of the weights. The absolute value of weight tends to be reduced. Some of them may be reduced to zeros. So, the only difference between Ridge and Lasso is that instead of taking the square of the coefficients, magnitudes are taken into account.

https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.Lasso.html

```python
from sklearn.linear_model import Lasso
```

```python
lasso_regr_model = Lasso(alpha=1)

lasso_regr_model.fit(X_train, y_train)
```

```python
lasso_regr_model.coef_
```

```python
lasso_regr_model.score(X_train, y_train)
```

```python
lasso_regr_model.score(X_test, y_test)
```

```python
cross_val_results(X, y, model=Lasso(alpha=1))
```

#### ElasticNet

ElasticNet is a hybrid of Lasso and Ridge, where both absolute value penalisation and squared penalisation are included, regulated by the L1_ratio coefficient. The L1_ratio takes values between 0 and 1. 

* L1_ratio = 0 is the Lasso penalty.
* L1_ratio = 1 is the Ridge penalty. 

L1_ratio between 0 and 1 is a combination of the Ridge and Lasso penalty. 

For Alpha = 0, it is equivalent to Linear Regression.

https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.ElasticNet.html

```python
from sklearn.linear_model import ElasticNet
```

```python
elastic_regr_model = ElasticNet(alpha=1, l1_ratio = 0.5)

elastic_regr_model.fit(X_train, y_train)

# alpha - penalty term
```

```python
elastic_regr_model.coef_
```

```python
elastic_regr_model.score(X_train, y_train)
```

```python
elastic_regr_model.score(X_test, y_test)
```

```python
cross_val_results(X, y, model=ElasticNet(alpha=1, l1_ratio = 0.5))
```

The regularisation helps us reduce multicollinearity by decreasing the magnitude of the coefficients. This way, the coefficients that harm the model will have a reduced influence on the target.

The benefit of regularisation over the VIF is that it automatically determines which coefficients harm the model and penalises them, but does not remove them all from the model. Thus, the score will not be significantly affected.

#### Task 1: Train a linear regression model using the Ridge, Lasso, and ElasticNet models.

Validate the created models and explain the results.

```python
# Load the cars dataset
cars = sns.load_dataset('mpg')

# Check dataset
cars.head()
```

```python
# Review and clean the dataset
cars.info()
```

```python
# Drop rows with missing values
cars = cars.dropna()
cars.info()
```

```python
# Define features and target variable
cars_X = cars[['cylinders', 'displacement', 'horsepower', 'weight', 'acceleration', 'model_year']]
cars_y = cars['mpg']

# Train-test split
cars_X_train, cars_X_test, cars_y_train, cars_y_test = train_test_split(cars_X, cars_y, test_size=0.2, random_state=42)
```

**Linear Regression**

```python
# Linear regression model
cars_model = LinearRegression()
cars_model.fit(cars_X_train, cars_y_train)

# Predict on the test set
cars_y_pred = cars_model.predict(cars_X_test)

# Store metrics
linear_r2_train = cars_model.score(cars_X_train, cars_y_train)
linear_r2_test = cars_model.score(cars_X_test, cars_y_test)
linear_mae = mean_absolute_error(cars_y_test, cars_y_pred)
linear_rmse = root_mean_squared_error(cars_y_test, cars_y_pred)

# Display results
print(f"Train R^2: {linear_r2_train}")
print(f"Test R^2: {linear_r2_test}")
print(f"MAE = {linear_mae}")
print(f"RMSE = {linear_rmse}\n")
print(f"Coefficients: {cars_model.coef_}")

linear_cv_scores = cross_val_results(cars_X, cars_y, model=LinearRegression())
```

**Ridge Regression**    

```python
# Ridge regression model
ridge_cars_model = Ridge(alpha=1)
ridge_cars_model.fit(cars_X_train, cars_y_train)

# Predict on the test set
ridge_cars_y_pred = ridge_cars_model.predict(cars_X_test)

# Store metrics
ridge_r2_train = ridge_cars_model.score(cars_X_train, cars_y_train)
ridge_r2_test = ridge_cars_model.score(cars_X_test, cars_y_test)
ridge_mae = mean_absolute_error(cars_y_test, ridge_cars_y_pred)
ridge_rmse = root_mean_squared_error(cars_y_test, ridge_cars_y_pred)

# Display results
print(f"Train R^2: {ridge_r2_train}")
print(f"Test R^2: {ridge_r2_test}")
print(f"MAE = {ridge_mae}")
print(f"RMSE = {ridge_rmse}\n")
print(f"Coefficients: {ridge_cars_model.coef_}")

ridge_cv_scores = cross_val_results(cars_X, cars_y, model=Ridge(alpha=1))
```

**Lasso Regression**

```python
# Lasso regression model
lasso_cars_model = Lasso(alpha=1)
lasso_cars_model.fit(cars_X_train, cars_y_train)

# Predict on the test set
lasso_cars_y_pred = lasso_cars_model.predict(cars_X_test)

# Store metrics
lasso_r2_train = lasso_cars_model.score(cars_X_train, cars_y_train)
lasso_r2_test = lasso_cars_model.score(cars_X_test, cars_y_test)
lasso_mae = mean_absolute_error(cars_y_test, lasso_cars_y_pred)
lasso_rmse = root_mean_squared_error(cars_y_test, lasso_cars_y_pred)

# Display results
print(f"Train R^2: {lasso_r2_train}")
print(f"Test R^2: {lasso_r2_test}")
print(f"MAE = {lasso_mae}")
print(f"RMSE = {lasso_rmse}\n")
print(f"Coefficients: {lasso_cars_model.coef_}")

lasso_cv_scores = cross_val_results(cars_X, cars_y, model=Lasso(alpha=1))
```

**Elastic Net**

```python
# ElasticNet regression model
elastic_cars_model = ElasticNet(alpha=1, l1_ratio=0.5)
elastic_cars_model.fit(cars_X_train, cars_y_train)

# Predict on the test set
elastic_cars_y_pred = elastic_cars_model.predict(cars_X_test)

# Store metrics
elastic_r2_train = elastic_cars_model.score(cars_X_train, cars_y_train)
elastic_r2_test = elastic_cars_model.score(cars_X_test, cars_y_test)
elastic_mae = mean_absolute_error(cars_y_test, elastic_cars_y_pred)
elastic_rmse = root_mean_squared_error(cars_y_test, elastic_cars_y_pred)

# Display results
print(f"Train R^2: {elastic_r2_train}")
print(f"Test R^2: {elastic_r2_test}")
print(f"MAE = {elastic_mae}")
print(f"RMSE = {elastic_rmse}\n")
print(f"Coefficients: {elastic_cars_model.coef_}")

elastic_cv_scores = cross_val_results(cars_X, cars_y, model=ElasticNet(alpha=1, l1_ratio=0.5))
```

**Compare Results for Linear, Ridge, Lasso and Elastic**

```python
# Performance comparison across all measures used
models = ['Linear Regression', 'Ridge Regression', 'Lasso Regression', 'ElasticNet Regression']

cv_scores = [linear_cv_scores, ridge_cv_scores, lasso_cv_scores, elastic_cv_scores]

performance_df = pd.DataFrame({
    'Model':    models,
    'R2 Train': [linear_r2_train, ridge_r2_train, lasso_r2_train, elastic_r2_train],
    'R2 Test':  [linear_r2_test,  ridge_r2_test,  lasso_r2_test,  elastic_r2_test],
    'MAE':      [linear_mae,      ridge_mae,      lasso_mae,      elastic_mae],
    'RMSE':     [linear_rmse,     ridge_rmse,     lasso_rmse,     elastic_rmse],
    'CV Mean':  [s.mean() for s in cv_scores],
    'CV Std':   [s.std()  for s in cv_scores],
})
print(performance_df)
```

**Results:**
All four models performed very similarly, with only small differences between them. Linear and Ridge Regression gave almost the same results, which suggests that Ridge's alpha=1 penalty isn't strong enough to have much effect — likely because multicollinearity isn't causing a big problem in this dataset.

Lasso and ElasticNet scored slightly lower R² and higher MAE/RMSE on unseen data. This makes sense as both models reduce or remove coefficients, which can hurt performance when the data doesn't have strong multicollinearity to correct for. That said, their CV Mean was very slightly higher (0.800 vs 0.799), so the difference is too small to matter.

Overall, regularisation doesn't appear to be necessary here — Linear Regression holds up just as well without it.


```python
# RIdge regression with different alpha values
alpha_values = [0.01, 0.1, 1, 10, 100]
ridge_cv_results = []
for alpha in alpha_values:
    print(f"Ridge Regression with alpha = {alpha}")
    scores = cross_val_results(cars_X, cars_y, model=Ridge(alpha=alpha))
    ridge_cv_results.append(scores)
    print("\n")
```

```python
# Lasso regression with different alpha values
lasso_cv_results = []
for alpha in alpha_values:
    print(f"Lasso Regression with alpha = {alpha}")
    scores = cross_val_results(cars_X, cars_y, model=Lasso(alpha=alpha))
    lasso_cv_results.append(scores)
    print("\n")
```

```python
# ElasticNet regression with different alpha values
elastic_cv_results = []
for alpha in alpha_values:
    print(f"ElasticNet Regression with alpha = {alpha}")
    scores = cross_val_results(cars_X, cars_y, model=ElasticNet(alpha=alpha, l1_ratio=0.5))
    elastic_cv_results.append(scores)
    print("\n")
```

```python
# Compare models with different alpha values
alpha_performance_df = pd.DataFrame({
    'Alpha': alpha_values,
    'Ridge CV Mean': [s.mean() for s in ridge_cv_results],
    'Ridge CV Std': [s.std() for s in ridge_cv_results],
    'Lasso CV Mean': [s.mean() for s in lasso_cv_results],
    'Lasso CV Std': [s.std() for s in lasso_cv_results],
    'ElasticNet CV Mean': [s.mean() for s in elastic_cv_results],
    'ElasticNet CV Std': [s.std() for s in elastic_cv_results],
})
print(alpha_performance_df)
```

**Conclusion**
Having reviewed the results using Alpha = 1, I re-ran the models across a range of alpha values to see if this would change anything.

Ridge stayed consistent across all alphas, which confirms it isn't sensitive to the penalty strength on this dataset. Lasso and ElasticNet were steady at lower alphas (0.01–1) but dropped off sharply at alpha=10 and above, as the penalty becomes too aggressive and removes coefficients that are actually useful.

Overall this still supports the original conclusion — Linear Regression is the best fit for this dataset, and regularisation doesn't add meaningful benefit. The likely reason is that the dataset does not have severe multicollinearity, so there is nothing for the penalty to correct. The alpha sweep also shows that if Lasso or ElasticNet were to be used, the alpha would need to be kept low (≤1) to avoid hurting performance.
