<h3> A Basic example </h3>

<h3> Pandas Series </h3>

- A pandas series is a 1-dimension array (a column)

```python
# Testing a Pandas Series
a = [1, 7, 2]
myvar2 = pd.Series(a)
print(myvar2)
```

- You can label a series using the index argument

```python
# Testing a Pandas Series with custom index
myvar3 = pd.Series(a, index = ["x", "y", "z"])
print(myvar3)
```

```python
print(myvar3["y"])
```

- A dictionary can be used for creating a series
- - The key values pairs act as a series label and value

```python
# Example of a Series created from a Dictionary
calories = {"day1": 420, "day2": 380, "day3": 390}
myvar4 = pd.Series(calories)
print(myvar4)
```

- You can access specific elements using the index argument

```python
# Using the index argument to select only some of the items in the dictionary as Series values
myvar5 = pd.Series(calories, index = ["day1", "day2"])
print(myvar5)
```

<h3> Pandas DataFrames </h3>

- Multidimension arrays are DataFrames in Pandas

```python
# Example of a Pandas DataFrame
data = {
    "calories": [420, 380, 390],
    "duration": [50, 40, 45]
}

mydf = pd.DataFrame(data)
print(mydf)
```

- The "loc" attribute  returns one or more rows

```python
# An example of locating a row in a DataFrame
print(mydf.loc[0])
```

```python
# An example of locating multiple rows in a DataFrame
print(mydf.loc[[0, 1]])
```

- You can also name your indexes

```python
# Example of naming the index in a DataFrame
mydf2 = pd.DataFrame(data, index = ["day1", "day2", "day3"])
print(mydf2)
```

- You can then also locate using the named index

```python
# Testing a Pandas DataFrame with custom index
mydf3 = pd.DataFrame(data, index = ["day1", "day2", "day3"])
print(mydf3)
```
